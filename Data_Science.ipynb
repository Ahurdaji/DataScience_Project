```python
sns.histplot(df["median_house_value"], bins=50)
plt.title("Distribution of House Prices"); plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap"); plt.show()
```

```python
sns.scatterplot(x="median_income", y="median_house_value", data=df, alpha=0.2)
plt.title("Income vs House Price"); plt.show()
```

```python
df.plot(kind="scatter", x="longitude", y="latitude", c="median_house_value",
        cmap="jet", alpha=0.4, figsize=(10,7), colorbar=True)
plt.title("House Prices by Location"); plt.show()
```

```python
sns.boxplot(x="ocean_proximity", y="median_house_value", data=df)
plt.title("House Price by Ocean Proximity"); plt.show()
```
